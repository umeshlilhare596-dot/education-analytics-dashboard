# Education Analytics Dashboard

**Purpose:** Analyze student marks, attendance, and academic performance using sample classroom data.

> Note: The dataset below is sample data for demonstrating the project. Replace it with real student data if needed.

```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Education Analytics Dataset (sample data)
data = {
    "Student": ["Amit", "Priya", "Rahul", "Sneha", "Rohan", "Pooja", "Vikas", "Neha", "Arjun", "Anjali"],
    "Maths": [85, 92, 67, 88, 76, 95, 58, 81, 73, 90],
    "Science": [80, 95, 70, 85, 78, 92, 60, 84, 75, 89],
    "English": [78, 90, 72, 86, 80, 94, 65, 82, 77, 91],
    "Attendance": [90, 98, 75, 92, 80, 99, 65, 88, 78, 95]
}

df = pd.DataFrame(data)
df["Total"] = df[["Maths", "Science", "English"]].sum(axis=1)
df["Average"] = df[["Maths", "Science", "English"]].mean(axis=1).round(2)
df["Result"] = df["Average"].apply(lambda x: "Pass" if x >= 35 else "Fail")

print("EDUCATION ANALYTICS DASHBOARD")
print("=" * 45)
print("Total Students:", len(df))
print("Average Class Marks:", round(df["Average"].mean(), 2))
print("Average Attendance:", round(df["Attendance"].mean(), 2), "%")
print("Students Passed:", int((df["Result"] == "Pass").sum()))
print("Students Failed:", int((df["Result"] == "Fail").sum()))

print("\nSTUDENT PERFORMANCE TABLE")
display(df)
```

```python
# Graph 1: Subject-wise average marks
subject_avg = df[["Maths", "Science", "English"]].mean()

plt.figure(figsize=(8, 4.5))
subject_avg.plot(kind="bar")
plt.title("Subject-wise Average Marks")
plt.ylabel("Average Marks")
plt.xlabel("Subject")
plt.ylim(0, 100)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Graph 2: Student attendance
plt.figure(figsize=(10, 5))
plt.bar(df["Student"], df["Attendance"])
plt.title("Student Attendance Analysis")
plt.ylabel("Attendance (%)")
plt.xlabel("Student")
plt.ylim(0, 100)
plt.xticks(rotation=35)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Top 3 students by average marks
top3 = df.nlargest(3, "Average")[["Student", "Average", "Attendance", "Result"]]
print("TOP 3 STUDENTS")
display(top3)
```
